## 10^5

```python
DATADIR = "/data2/yujiehe/rich-monitor/YujieSnellius/R0.47M0.5BH100000beta1S60n1.5ComptonHiResNewAMR"
snapnum=156
SNAPDIR = os.path.join(DATADIR, f"snap_full_{snapnum}.h5")
snap = richio.load(SNAPDIR)
print(snapnum, snap.time.in_units("day"))
alpha_rosseland = richio.opacity.rosseland_alpha(snap.T, snap.rho)

snap.plots.projection(
    alpha_rosseland,
    res=512,
    box_size=[-100, -100, -100, 100, 100, 100],
    colorbar_label=r"$\int\,\alpha_\mathrm{ross}dz$",
    vmin=-1,
    vmax=2,
    cmap="Blues",
)
plt.show()
```
